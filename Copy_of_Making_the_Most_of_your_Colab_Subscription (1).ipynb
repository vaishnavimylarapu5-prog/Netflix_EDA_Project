```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("libraries imported sucessfully")
```

```
df=pd.read_csv("netflix_titles.csv")
df.head()
```

```
df.info()
```

```
df.isnull().sum()
```

```
df.duplicated().sum()
```

```
#check missing values
df.isnull().sum()

#remove duplicated rows
df.drop_duplicates(inplace=True)

#fill missing values
df['director']=df['director'].fillna('unknown')
df['cast']=df['cast'].fillna('not available')
df['country']=df['country'].fillna('unknown')
df['rating']=df['rating'].fillna(df['rating'].mode()[0])

#remove rows with missing title or date_added
df.dropna(subset=['title','date_added'],inplace=True)

#check missing values again
df.isnull().sum()
```

```
plt.figure(figsize=(6,4))
sns.countplot(x='type',data=df)
plt.title("movies vs tv shows")
plt.show()
```

```
year_count=df['release_year'].value_counts().sort_index()
plt.figure(figsize=(10,5))
plt.plot(year_count.index,year_count.values)
plt.title("content realesed by year")
plt.xlabel("year")
plt.ylabel("number of titles")
plt.show()
```

```
plt.figure(figsize=(8,5))
plt.hist(df['release_year'],bins=20)
plt.title("distribution of release years")
plt.xlabel("release year")
plt.ylabel("frequency")
plt.show()
```

```
plt.figure(figsize=(8,5))
plt.scatter(df.index,df['release_year'])
plt.title("release year vs dataset index")
plt.xlabel("dataset index")
plt.ylabel("release year")
plt.show()
```

```
df['type'].value_counts().plot(kind='pie',autopct='%1.1f%%',figsize=(6,6))
plt.title("movies and tv shows distribution")
plt.ylabel("")
plt.show()
```
